```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

s = Service('d:/chromedriver-win64/chromedriver.exe')

driver = webdriver.Chrome(service=s)
driver.get('https://www.smartprix.com/mobiles')

driver.find_element(by=By.XPATH , value='//*[@id="app"]/main/aside/div/div[5]/div[2]/label[1]/input').click()
driver.find_element(by=By.XPATH , value='//*[@id="app"]/main/aside/div/div[5]/div[2]/label[2]/input').click()

old_height = driver.execute_script("return document.body.scrollHeight")
while True:

    clicked = driver.execute_script("""
        let elements = document.getElementsByTagName('*');

        for (let el of elements) {
            if (el.innerText && el.innerText.trim() === 'Load More') {
                el.scrollIntoView({block: 'center'});
                el.click();
                return true;
            }
        }

        return false;
    """)

    if not clicked:
        print("Load More not found")
        break

    time.sleep(2)

    new_height = driver.execute_script(
        "return document.body.scrollHeight"
    )

    print(old_height)
    print(new_height)

    if new_height == old_height:
        time.sleep(2)

        new_height = driver.execute_script(
            "return document.body.scrollHeight"
        )

        if new_height == old_height:
            break

    old_height = new_height

html = driver.page_source

with open('smartprix.html','w',encoding='utf-8') as f:
    f.write(html)    
```

```python
with open('smartprix.html','r',encoding='utf-8') as f:
    html = f.read()
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(html,'html.parser')
```

```python
containers = soup.find_all('div',{'class':"sm-product has-tag has-features has-actions"})
```

```python
import re
import numpy as np
import pandas as pd
```

```python
name = []
price = []
ratings = []
sim = []
processor = []
ram = []
battery = []
display = []
camera = []
card = []
os = []

for i in containers:

    # NAME
    try:
        name.append(i.find('h2').get_text(strip=True))
    except:
        name.append(np.nan)

    # PRICE
    try:
        price.append(i.find('span', {'class': 'price'}).get_text(strip=True))
    except:
        price.append(np.nan)

    # SPEC SCORE
    try:
        score = i.select_one("div.score")
        number = re.search(r'\d+', score.get_text())
        ratings.append(int(number.group()) if number else np.nan)
    except:
        ratings.append(np.nan)

    # SPECS
    try:
        specs = i.select("ul.sm-feat.specs li")

        # SIM
        try:
            sim.append(specs[0].get_text(" ", strip=True))
        except:
            sim.append(np.nan)

        # PROCESSOR
        try:
            processor.append(specs[1].get_text(" ", strip=True))
        except:
            processor.append(np.nan)

        # RAM
        try:
            ram.append(specs[2].get_text(" ", strip=True))
        except:
            ram.append(np.nan)

        # BATTERY
        try:
            battery.append(specs[3].get_text(" ", strip=True))
        except:
            battery.append(np.nan)

        # DISPLAY
        try:
            display.append(specs[4].get_text(" ", strip=True))
        except:
            display.append(np.nan)

        # CAMERA
        try:
            camera.append(specs[5].get_text(" ", strip=True))
        except:
            camera.append(np.nan)

        # MEMORY CARD
        try:
            card.append(specs[6].get_text(" ", strip=True))
        except:
            card.append(np.nan)

        # OS
        try:
            os.append(specs[7].get_text(" ", strip=True))
        except:
            os.append(np.nan)

    except:
        sim.append(np.nan)
        processor.append(np.nan)
        ram.append(np.nan)
        battery.append(np.nan)
        display.append(np.nan)
        camera.append(np.nan)
        card.append(np.nan)
        os.append(np.nan)
```

```python
df = pd.DataFrame({
    'Name': name,
    'Price': price,
    'Spec Score': ratings,
    'SIM': sim,
    'Processor': processor,
    'RAM': ram,
    'Battery': battery,
    'Display': display,
    'Camera': camera,
    'Memory Card': card,
    'OS': os
})
```
