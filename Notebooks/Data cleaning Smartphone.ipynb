## Data Assessing

`NOTE :-` All ROWS NO. ARE BASED ON PHASEONE DATASET 

`Quality Issues`
 
1. **model** - some brands are written differently like OPPO in model col.
2. **price** - has unneccesary `rupees` symbol
3. **price** - has " , " between numbers
4. **row no.** - 366(Jio Bharat J1 4G)phone has missing processor value , therefore all data is mixed up of that row. there are several phones like this
                 so we drop those phones whose price is below 3400 as below that all are feature phones not smartphones. ~ roughly 80+ phones
5. `Feature Phones` - there are many feature phones `drop` them        
6. **ram** - row no.(276(Apple iPhone 16 Plus), 478(Apple iPhone 16 Plus (256GB))
