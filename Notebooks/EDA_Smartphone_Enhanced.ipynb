```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler

# 1. Fill Processor Speed
features_cpu = [
    'Spec Score',
    'Number of Cores',
    'RAM Capacity',
    'ROM Capacity',
    'Processor Speed'
]

scaler_cpu = StandardScaler()
scaled_cpu = scaler_cpu.fit_transform(df[features_cpu])

imputer_cpu = KNNImputer(n_neighbors=5, weights='distance')
imputed_cpu = imputer_cpu.fit_transform(scaled_cpu)

imputed_cpu = scaler_cpu.inverse_transform(imputed_cpu)

mask_cpu = df['Processor Speed'].isna()
df.loc[mask_cpu, 'Processor Speed'] = imputed_cpu[mask_cpu, -1]


# 2. Fill Fast Charging
features_charging = [
    'Spec Score',
    'Battery Capacity',
    'RAM Capacity',
    'ROM Capacity',
    'Fast Charging'
]

scaler_charging = StandardScaler()
scaled_charging = scaler_charging.fit_transform(df[features_charging])

imputer_charging = KNNImputer(n_neighbors=5, weights='distance')
imputed_charging = imputer_charging.fit_transform(scaled_charging)

imputed_charging = scaler_charging.inverse_transform(imputed_charging)

mask_charging = df['Fast Charging'].isna()
df.loc[mask_charging, 'Fast Charging'] = imputed_charging[mask_charging, -1]


# 3. Verify
print(df[['Processor Speed', 'Fast Charging']].isna().sum())
```

## Additional EDA: Display and Price

The following sections add a few simple analyses and features that may be useful for understanding smartphone prices and preparing a future price-prediction model. The code uses basic Pandas, Seaborn and Matplotlib.

### 1. Check the resolution values

Before creating categories, inspect the values that actually occur in the dataset. This helps us avoid assigning labels without checking the data format.

```python
df['Resolution'].value_counts().head(15)
```

### 2. Create resolution categories

This groups displays by their approximate pixel dimensions. Since smartphone resolutions use different aspect ratios, the category is based on the shorter and longer dimensions rather than requiring an exact resolution match. These are practical EDA categories, not necessarily the phone manufacturer's marketing labels.

```python
# Split resolution into two numbers, after removing extra spaces
resolution = df['Resolution'].str.strip().str.extract(r'(\d+)\s*x\s*(\d+)')

df['Resolution_1'] = pd.to_numeric(resolution[0], errors='coerce')
df['Resolution_2'] = pd.to_numeric(resolution[1], errors='coerce')

# Shorter and longer sides work for either width x height order
df['Short_Side'] = df[['Resolution_1', 'Resolution_2']].min(axis=1)
df['Long_Side'] = df[['Resolution_1', 'Resolution_2']].max(axis=1)

def resolution_category(row):
    short_side = row['Short_Side']
    long_side = row['Long_Side']

    if pd.isna(short_side) or pd.isna(long_side):
        return 'Unknown'
    elif short_side >= 2000 or long_side >= 3800:
        return 'Ultra HD (4K)'
    elif short_side >= 1400:
        return 'QHD / QHD+'
    elif short_side >= 1000:
        return 'FHD / FHD+'
    elif short_side >= 800:
        return 'Other'
    elif long_side >= 1500:
        return 'HD+'
    elif short_side >= 700:
        return 'HD'
    else:
        return 'Other'

df['Resolution_Category'] = df.apply(resolution_category, axis=1)

df['Resolution_Category'].value_counts()
```

```python
sns.countplot(data=df, x='Resolution_Category',
              order=df['Resolution_Category'].value_counts().index)
plt.xticks(rotation=30)
plt.title('Number of Phones by Resolution Category')
plt.show()
```

**Conclusion:** Use the chart to see which resolution groups are most common in this dataset. The categories are approximate and should be described as dataset-based groups rather than official marketing labels.

### 3. Calculate PPI (pixel density)

PPI estimates how many pixels fit into one inch of the display. A higher PPI generally means a sharper display, although panel quality also matters.

```python
df['PPI'] = (
    (df['Resolution_1']**2 + df['Resolution_2']**2)**0.5
    / df['Screen Size']
)

sns.scatterplot(data=df, x='PPI', y='Price')
plt.title('Pixel Density vs Price')
plt.show()
```

**Conclusion:** Look for whether higher PPI values tend to appear at higher prices. A scatter plot shows association, not proof that PPI causes a higher price.

### 4. Calculate aspect ratio

Aspect ratio compares the longer side of the display with the shorter side. It helps compare screen shapes even when the exact resolutions differ.

```python
df['Aspect_Ratio'] = df['Long_Side'] / df['Short_Side']

sns.scatterplot(data=df, x='Aspect_Ratio', y='Price')
plt.title('Display Aspect Ratio vs Price')
plt.show()
```

### 7. Create simple specification features

These features are easy to interpret and may be useful for later modelling. They are derived from existing columns, so they do not introduce new external information.

```python
# Total number of cameras
df['Total_Cameras'] = (
    df['Number of Rear Cameras'] + df['Number of Front Cameras']
)

# Group battery capacity into simple tiers
df['Battery_Tier'] = pd.cut(
    df['Battery Capacity'],
    bins=[0, 4000, 5000, 6000, float('inf')],
    labels=['Small', 'Medium', 'Large', 'Very Large']
)

# Group refresh rates
df['Refresh_Rate_Tier'] = pd.cut(
    df['Refresh Rate'],
    bins=[0, 60, 90, 120, float('inf')],
    labels=['Up to 60Hz', '61-90Hz', '91-120Hz', 'Above 120Hz']
)

df[['Total_Cameras', 'Battery_Tier', 'Refresh_Rate_Tier']].head()
```

## Conclusions:

 - 1. **Price distribution** 
   The dataset contains 923 smartphones, with prices ranging from ₹4,499 to ₹4,49,900. The median price is ₹29,999, while the mean is approximately ₹43,191. The mean being substantially higher than the median, along with a positive skewness of 3.67, indicates that a relatively small number of expensive smartphones pull the average price upward. Therefore, the median is a more representative measure of the typical phone price in this dataset.

 - 2. **Expensive smartphones and outliers**
   There are 14 smartphones priced above ₹2,00,000 in the saved output. These include premium Apple models, Samsung foldable phones, and the Vivo X Fold 6. Such observations contribute to the right-skewed price distribution. They should not automatically be removed as errors because they may represent genuine premium products.

 - 3. **Spec Score distribution**
   The Spec Score ranges from 17 to 98, with a mean of approximately 64.49 and a median of 64. Its skewness is approximately -0.12, indicating that the distribution is relatively balanced compared with the highly right-skewed Price column. This suggests that the dataset contains smartphones across a broad range of specification scores without the same degree of extreme skew seen in pricing.  

 - 4. **Resolution distribution**
   The most common resolution is 1080 × 2400, found in 118 phones, followed by 720 × 1600 in 113 phones and 1080 × 2340 in 94 phones. These three resolutions alone account for 325 records, approximately 35.2% of the dataset. This shows that FHD-class and HD-class resolutions are both strongly represented in the smartphone sample.  

 - 5. `Correlation with Price`
   Processor Speed and ROM Capacity have the strongest positive linear relationships with Price, both with correlations of approximately 0.79. Spec Score (0.66) and RAM Capacity (0.61) also show substantial positive relationships. This suggests that processing performance, internal storage, and overall specifications are important candidates for further price-prediction analysis.  

    In comparison, Fast Charging (0.25), Has 5G (0.26), and Screen Size (0.30) have weaker positive relationships. Battery Capacity has a weak negative correlation (-0.11), indicating almost no linear relationship with price in this sample. 

    Extended Memory Available (-0.46) and Aspect Ratio (-0.49) have negative correlations. These should not be interpreted as evidence that expandable storage or a particular display shape reduces price; the results may reflect differences between premium and budget phone designs.

 - 6. **Processor Speed**
   Processor Speed has a strong positive correlation with Price (0.788) and an even stronger correlation with Spec Score (0.826). This indicates that faster processors tend to occur in phones with higher specification scores and higher prices. Processor Speed is therefore a promising feature to evaluate in a future price-prediction model, although its independent contribution should be tested alongside RAM, storage, and brand.

 - 7. **Missing-value analysis**
   The saved output shows 110 missing values in Extended Memory, while the other columns have no remaining pandas null values. The missing Extended Memory values represent approximately 11.9% of the 923 records. These values should be handled before machine learning, either through an appropriate imputation method or a missing-value indicator, depending on what the field represents.

 - 8. **Resolution and PPI feature engineering**
   The Resolution column was separated into two numerical dimensions, allowing display features such as total pixel count, PPI, and aspect ratio to be calculated. The resulting correlation analysis shows that PPI has a positive correlation of approximately 0.446 with Price. This is a moderate relationship, suggesting that pixel density may provide useful information for price prediction, although it is weaker than the relationships observed for Processor Speed and ROM Capacity.

 - 9. **RAM and storage**
   RAM Capacity has a positive correlation of 0.611 with Price, while ROM Capacity has a stronger correlation of 0.788. This indicates that higher storage configurations are more strongly associated with higher prices than RAM capacity in this dataset. Both features are useful candidates for machine learning, but their relationships may overlap with processor performance and brand positioning. 

 - 10. **Before and after imputation**
   After KNN imputation, the correlation between Price and Processor Speed changed from 0.80 to 0.79, while the correlation between Price and Fast Charging decreased from 0.33 to 0.25. The Processor Speed relationship remained almost unchanged, whereas the Fast Charging relationship was more affected by imputation.
